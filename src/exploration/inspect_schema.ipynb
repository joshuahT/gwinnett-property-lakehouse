```python
import io
import zipfile
import requests
import pandas as pd

url = "https://gwinnettcounty.com/content/departments/its/gis-maps/tax_assessor_property_ownership_data.zip"

resp = requests.get(url, timeout=180)
resp.raise_for_status()

with zipfile.ZipFile(io.BytesIO(resp.content)) as zf:
    xlsx_files = [n for n in zf.namelist() if n.lower().endswith(".xlsx")]
    print(f"Found {len(xlsx_files)} file(s): {xlsx_files}")

    for name in xlsx_files:
        with zf.open(name) as f:
            df = pd.read_excel(f, engine="openpyxl", nrows=20)  # peek only
        print(f"\n=== {name} ===")
        print(list(df.columns))
        display(df.head())
```
